```python
import xarray as xr
import os
import numpy as np
```

```python
start_dir = os.getcwd()
watershed = "watersheds/huc8_07120004/"
resolution = 30
event_date = '2013-04-10_2013-04-23'
```

```python
# Build file paths
file_path = os.path.join(
    start_dir, watershed, f"data_for_HiPIMS_{resolution}",
    "precipitation_st4", f"precipitation_event_{event_date}.nc"
)

adjust_file_path = os.path.join(
    start_dir, watershed, f"data_for_HiPIMS_{resolution}",
    "precipitation_st4", f"precipitation_event_{event_date}_adjusted.nc"
)

# Open dataset
ds = xr.open_dataset(file_path)

# Select variable and slice dimensions
pr_fixed = ds["prcp"].isel(
    time=slice(175, 175 + 96 + 1),  # 97 timesteps
    y=slice(0, 39),
    x=slice(0, 13)
)

# Convert back to Dataset (keeps only sliced coordinates)
fixed_ds = pr_fixed.to_dataset(name="prcp")

# Save new NetCDF (overwrite if exists)
fixed_ds.to_netcdf(adjust_file_path, mode="w")

# Optional sanity check
print(fixed_ds.sizes)
```
